```python
def run_inference(image_path,model):
  results = model.predict(source = image_path, imgsz=(1025))
  plot = results[0].plot()
  plot = cv2.cvtColor(plot, cv2.COLOR_BGR2RGB)
  plt.figure(figsize=(5, 5))
  plt.imshow(plot)
  plt.axis('off')
  plt.show()
```
