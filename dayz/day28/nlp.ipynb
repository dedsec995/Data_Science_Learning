### Semenatic Similarity Scoring

```python
def cos_sim(A,B):
    return np.inner(A,B)/(norm(A)*norm(B))

def euclidean(A,B):
    return norm(A-B)
```

```python
p1 = np.array((1, 2, 3))
p2 = np.array((1, 1, 1))
print(euclidean(p1,p2))
print(cos_sim(p1,p2))
```

```python
def is_it_sim(textA, textB, thresh=0.28, sim_func = cos_sim):
    embeddingsA = embed([textA])
    embeddingsB = embed([textB])
    sim_score = sim_func(embeddingsA,embeddingsB)
    if sim_score > thresh:
        print("They are Similar!")
    else:
        print("Not at All...")
    return sim_score, sim_score>thresh
```

```python
QuestionA = "This is a technology Company that builds computers"
AnswerA = "I had a delicious fruit called apple"
AnswerB = "I am writing a program on my Apple computer"

print(is_it_sim(QuestionA,AnswerB,sim_func=euclidean))
```
