## Any number and returns prime factors

```python
def my_prime(x):
    listy = []
    for i in range(2,x):
        if x%i == 0:
            listy.append(i)
    listy = [x for x in listy if is_prime(x)]
    return listy
def is_prime(x):
    if x < 2:
        return False
    for i in range(2, int(x ** 0.5) + 1):
        if x % i == 0:
            return False
    return True    
```

```python
def their_primee(x):
    factors = []
    div_it = 2
    while x >= 2:
        if x % div_it == 0:
            factors.append(div_it)
            x //= div_it
        else:
            div_it += 1
    return factors

print('*'*40)
x = 1257
print(my_prime(x))
print()
print(their_primee(x))
print('*'*40)
x = 256
print(my_prime(x))
print()
print(their_primee(x))
print('*'*40)
```

# Rotate the Matrix

```python
def rotate90(matrix):
    n = len(matrix)
    for i in range(n):
        for j in range(i, n):
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    # print(matrix)
    for i in range(n):
        matrix[i].reverse()
    return matrix

def rotate_it(matrix, n):
    n = n % 4 
    for _ in range(n):
        rotate90(matrix)
    return matrix
matrix = [[1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]]
rotate_it(matrix,1)
```
